```python
import pandas as pd 
```

```python
grouped_df = pd.read_csv("../data/data_with_groups.csv")
gen_df = pd.read_csv("../data/investigator_nacc72.csv")
biom_df = pd.read_csv("../data/investigator_fcsf_nacc72.csv")
```

```python
biom_merged_df = pd.merge(biom_df, grouped_df, on="NACCID")
```

```python
print(biom_merged_df.columns)
biom_merged_df
```

```python
# Filtere gen_df vorab auf Einträge, die eine Gruppe haben
gen_with_group = gen_df[gen_df["Group"].notna()]

# Jetzt der Merge (nur noch Übereinstimmungen bleiben übrig)
df_biom_gen_group = pd.merge(biom_df, gen_with_group, on="NACCID", how="inner")
```

```python
df_biom_gen_group.to_csv("../data/df_biogen.csv", index = False)
```
